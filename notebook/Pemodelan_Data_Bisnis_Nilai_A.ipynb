```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load data
df = pd.read_excel('/content/data persebaran industri.xlsx')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
# Memilih kolom yang akan digunakan
cols_used = [
    "Nama Perusahaan",
    "Uraian Jenis Perusahaan",
    "Uraian Risiko Proyek",
    "Uraian Skala Usaha",
    "Judul Kbli",
    "luas_tanah",
    "Jumlah Investasi",
    "TKI",
    "kecamatan_usaha"
]

df1 = df[cols_used].copy()
print("Shape setelah seleksi kolom:", df1.shape)
df1.head()
```

```python
df1.dtypes
```

```python
df1.info()
```

```python
# Standarisasi uraian jenis perusahaan
jenis_map = {
    "Perorangan": "Perorangan",
    "Perseroan Terbatas (PT)": "PT",
    "Perseroan Terbatas (PT) Perorangan": "PT Perorangan",
    "Persekutuan Komanditer (CV / Commanditaire Vennootschap)": "CV",
    "Persekutuan Firma (Fa / Venootschap Onder Firma)": "Firma",
    "Persekutuan dan Perkumpulan": "Persekutuan dan Perkumpulan",
    "Badan Hukum Lainnya": "Badan Hukum Lainnya",
    "Badan Layanan Umum (BLU)": "BLU",
    "Perusahaan Umum Daerah (Perumda)": "Perumda"
}

df1["Uraian Jenis Perusahaan"] = df1["Uraian Jenis Perusahaan"].map(jenis_map)

df1["Uraian Jenis Perusahaan"].value_counts(dropna=False)
```

```python
# Filter skala usaha hanya IKM
df1["Uraian Skala Usaha"].value_counts()
```

```python
df1 = df1[df1["Uraian Skala Usaha"].isin([
    "Usaha Mikro",
    "Usaha Kecil",
    "Usaha Menengah"
])].copy()

print("Shape setelah filter IKM:", df1.shape)
df1["Uraian Skala Usaha"].value_counts(normalize=True) * 100
```

```python
# Cek Missing Value (SETELAH FILTER & MAPPING)
df1.isnull().sum().sort_values(ascending=False)
```

```python
# Drop missing yang tidak sah diimpute
df1 = df1.dropna(subset=[
    "Uraian Jenis Perusahaan",
    "Uraian Skala Usaha",
    "Jumlah Investasi",
    "TKI"
])

print("Shape setelah drop missing utama:", df1.shape)
```

```python
# Pastikan & Konversi Variabel Numerik
# Bersihkan format angka (jika masih string)
df1["Jumlah Investasi"] = (
    df1["Jumlah Investasi"]
    .astype(str)
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False)
)

df1["Jumlah Investasi"] = pd.to_numeric(df1["Jumlah Investasi"], errors="coerce")
df1["luas_tanah"] = pd.to_numeric(df1["luas_tanah"], errors="coerce")
df1["TKI"] = pd.to_numeric(df1["TKI"], errors="coerce")

# cek ulang tipe data
df1.dtypes
```

```python
# cek missing value final
df1.isnull().sum().sort_values(ascending=False)
```

```python
df1 = df1.dropna(subset=["luas_tanah"])

print("Shape setelah drop missing luas_tanah:", df1.shape)
```

```python
df1.isnull().sum().sort_values(ascending=False)
```

```python
# Statistik deskriptif numerik
num_cols = ["Jumlah Investasi", "luas_tanah", "TKI"]
df1[num_cols].describe()
```

```python
# Distribusi variabel numerik
df1[num_cols].hist(bins=30, figsize=(12,4))
plt.tight_layout()
plt.show()
```

```python
# Heatmap korelasi antar variabel numerik
plt.figure(figsize=(6,4))

corr = df1[num_cols].corr(method="pearson")

sns.heatmap(
    corr,
    annot=True,
    cmap="YlGnBu",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Heatmap Korelasi Antar Variabel Numerik")
plt.show()
```

```python
# Distribusi skala usaha
df1["Uraian Skala Usaha"].value_counts()
df1["Uraian Skala Usaha"].value_counts(normalize=True) * 100
```

```python
# Distribusi jenis perusahaan
df1["Uraian Jenis Perusahaan"].value_counts()
df1["Uraian Jenis Perusahaan"].value_counts(normalize=True) * 100
```

```python
df_context = df.loc[df1.index, [
    "Nama Perusahaan",
    "kecamatan_usaha",
    "Judul Kbli",
    "Uraian Jenis Perusahaan",
    "Uraian Risiko Proyek",
    "Jumlah Investasi",
    "TKI",
    "luas_tanah"
]].copy()
```

```python
# Target encoding
# target
y = df1["Uraian Skala Usaha"]

# cek kelas
y.value_counts()
```

```python
# fitur x
X = df1.drop(columns=["Uraian Skala Usaha"])
```

```python
# pisahkan target dan fitur
y = df1["Uraian Skala Usaha"]
X = df1.drop(columns=["Uraian Skala Usaha"])
```

```python
# Tentukan kategorikal dan dan numerikal
cat_cols = [
    "Uraian Jenis Perusahaan",
    "Uraian Risiko Proyek",
    "Judul Kbli",
    "kecamatan_usaha"
]

num_cols = ["Jumlah Investasi", "luas_tanah", "TKI"]
```

```python
# One Hot Encoding
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    drop="first",
    sparse_output=False,
    handle_unknown="ignore"
)

X_cat = encoder.fit_transform(X[cat_cols])
X_num = X[num_cols].values

X_final = np.hstack([X_num, X_cat])

print("Shape X_final:", X_final.shape)
```

```python
# Train test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_final,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Train:", X_train.shape)
print("Test :", X_test.shape)
```

```python
y_train.value_counts(normalize=True)
```

```python
y_test.value_counts(normalize=True)
```

```python
# Modelling
# Naive Bayes
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

nb = GaussianNB()
nb.fit(X_train, y_train)

y_pred_nb = nb.predict(X_test)

acc_nb = accuracy_score(y_test, y_pred_nb)
print("Accuracy Naive Bayes:", acc_nb)

print(classification_report(y_test, y_pred_nb))
```

```python
# Decision Tree
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(
    max_depth=10,
    min_samples_leaf=50,
    random_state=42
)

dt.fit(X_train, y_train)

y_pred_dt = dt.predict(X_test)

acc_dt = accuracy_score(y_test, y_pred_dt)
print("Accuracy Decision Tree:", acc_dt)

print(classification_report(y_test, y_pred_dt))
```

```python
# Random Forest
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    min_samples_leaf=20,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)

acc_rf = accuracy_score(y_test, y_pred_rf)
print("Accuracy Random Forest:", acc_rf)

print(classification_report(y_test, y_pred_rf))
```

```python
# Perbandingan akurasi antar model
acc_df = pd.DataFrame({
    "Model": ["Naive Bayes", "Decision Tree", "Random Forest"],
    "Accuracy": [acc_nb, acc_dt, acc_rf]
})

plt.figure(figsize=(6,4))

colors = ["#4C72B0", "#DD8452", "#55A868"]
bars = plt.bar(acc_df["Model"], acc_df["Accuracy"], color=colors)

# Grid sumbu Y
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.ylim(0.85, 0.92)
plt.title("Perbandingan Akurasi Model")
plt.ylabel("Accuracy")
plt.xlabel("Model")

# Angka di atas tiap batang
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.002,
        f"{height:.3f}",
        ha="center",
        va="bottom",
        fontsize=10
    )

plt.tight_layout()
plt.show()
```

```python
# Menetapkan model final
model_final = dt
```

```python
# Feature Importance
feature_names = (
    ["Jumlah Investasi", "luas_tanah", "TKI"] +
    list(encoder.get_feature_names_out(cat_cols))
)

importance_df = pd.DataFrame({
    "feature": feature_names,
    "importance": model_final.feature_importances_
}).sort_values(by="importance", ascending=False)

importance_df.head(7)
```

```python
# Prediksi Skala Usaha pada Data Uji
y_pred_dt = model_final.predict(X_test)
```

```python
# Menggabungkan dengan data intefier
hasil_prediksi_lengkap = pd.DataFrame({
    "Nama Perusahaan": df_context.loc[y_test.index, "Nama Perusahaan"],
    "Kecamatan": df_context.loc[y_test.index, "kecamatan_usaha"],
    "Judul KBLI": df_context.loc[y_test.index, "Judul Kbli"],
    "Jenis Perusahaan": df_context.loc[y_test.index, "Uraian Jenis Perusahaan"],
    "Risiko Proyek": df_context.loc[y_test.index, "Uraian Risiko Proyek"],
    "Jumlah Investasi": df_context.loc[y_test.index, "Jumlah Investasi"],
    "TKI": df_context.loc[y_test.index, "TKI"],
    "Luas Tanah": df_context.loc[y_test.index, "luas_tanah"],
    "Skala Usaha Aktual": y_test.values,
    "Skala Usaha Prediksi": y_pred_dt
})

hasil_prediksi_lengkap.head()
```

```python
# Definisi Usaha PRIORITAS PEMBINAAN
prioritas_pembinaan_lengkap = hasil_prediksi_lengkap[
    (hasil_prediksi_lengkap["Skala Usaha Aktual"] == "Usaha Mikro") &
    (hasil_prediksi_lengkap["Skala Usaha Prediksi"] != "Usaha Mikro")
].copy()

prioritas_pembinaan_lengkap.head()
```

```python
# Ringkasan Prioritas Pembinaan per Kecamatan
prioritas_kecamatan = (
    prioritas_pembinaan_lengkap
    .groupby("Kecamatan")
    .size()
    .reset_index(name="Jumlah Usaha Prioritas")
    .sort_values(by="Jumlah Usaha Prioritas", ascending=False)
)

prioritas_kecamatan.head(10)
```

```python
#Rekomendasi Pembinaan
def rekomendasi(row):
    if row["Skala Usaha Prediksi"] == "Usaha Kecil":
        return "Pendampingan peningkatan kapasitas produksi dan kepatuhan perizinan"
    elif row["Skala Usaha Prediksi"] == "Usaha Menengah":
        return "Pembinaan lanjutan: sertifikasi, akses pembiayaan, dan ekspansi"
    else:
        return "Pembinaan dasar"

prioritas_pembinaan_lengkap["Rekomendasi Pembinaan"] = (
    prioritas_pembinaan_lengkap.apply(rekomendasi, axis=1)
)

prioritas_pembinaan_lengkap.head()
```

```python
hasil_prediksi_lengkap.to_excel("hasil_klasifikasi_skala_usaha.xlsx", index=False)
prioritas_pembinaan_lengkap.to_excel("prioritas_pembinaan_ikm.xlsx", index=False)
prioritas_kecamatan.to_excel("prioritas_kecamatan_pembinaan.xlsx", index=False)
```
